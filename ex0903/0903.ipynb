### 2. 문자열 다루기

```python
# 문자열도 for 문으로 한 글자씩 꺼낼 수 있다
for x in "banana":
  print(x)
```

```python
# len(): 문자열 길이 (공백과 느낌표도 한 글자로 세서 12)
a = "Hello World!"
print(len(a))

for x in a:
  print(x)
```

```python
# ⚠️ 변수 이름을 str로 지으면 내장 함수 str()을 덮어쓴다.
# 이 셀을 실행한 뒤에는 같은 노트북에서 str(3) 같은 코드가 'str' object is not callable 오류를 낸다.
# text처럼 다른 이름을 쓰자.
str = "Hello, World!"
print(len(str))
print()

for x in str:
  print(x)
  print("---")
print("*****")
```

```python
# in: 문자열 안에 포함되어 있는지 (True/False)
txt = "The best things in life are free!"
if "free" in txt:
  print("Yes, 'free' is present.")
```

```python
# not in: 포함되어 있지 않은지
txt = "The best things in life are free!"
print("expensive" not in txt)
```

```python
# 슬라이싱 [시작:끝] - 시작 위치는 포함, 끝 위치는 포함하지 않는다 (인덱스는 0부터)
# [:5]는 처음부터, [2:]는 끝까지
b = "Hello, World"
print(b[2:5])
print(b[:5])
print(b[2:])
```

```python
# 음수 인덱스는 뒤에서부터 센다 (-1이 마지막 글자). [-5:-2] -> 'Wor'
print(b[-5:-2])
```

```python
# upper(): 모두 대문자로
a = "Hello, World!"
print(a.upper())
```

```python
# lower(): 모두 소문자로
a = "Hello, World!"
print(a.lower())
```

```python
# strip(): 앞뒤 공백 제거
a = " Hello, World! "
print(a.strip())
```

```python
# replace(바꿀 것, 새 것)
a = "Hello, World!"
print(a.replace("H", "J"))
```

```python
# split(구분자): 구분자로 잘라서 리스트로
a = "Hello, World!"
print(a.split(","))
```

```python
# + 로 문자열 연결 (공백 없이 붙는다)
a = "Hello"
b = "World"
c = a + b
print(c)
```

```python
# 사이에 공백을 넣으려면 " "도 같이 더한다
a = "Hello"
b = "World"
c = a + " " + b
print(c)
```

```python
# f-string: 문자열 앞에 f를 붙이고 {변수}로 값을 끼워 넣는다
age = 36
txt = f"My name is John, I am {age}"
print(txt)
```

### 3. 클래스와 객체

```python
# 클래스 변수: 클래스에 직접 적은 변수는 모든 객체가 같은 값을 공유한다
class MyClass:
  x = 5
```

```python
# 클래스로 객체(인스턴스)를 만들고 속성에 접근
p1 = MyClass()
print(p1.x)
```

```python
# del: 객체(변수) 삭제
del p1
```

```python
# 같은 클래스로 만든 객체들은 클래스 변수 x를 같이 쓴다
p1 = MyClass()
p2 = MyClass()
p3 = MyClass()

print(p1.x)
print(p2.x)
print(p3.x)
```

```python
# __init__: 객체를 만들 때 자동으로 실행되는 함수(생성자)
# self.속성 = 값 으로 객체마다 다른 값을 저장한다
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

```python
# 같은 클래스로 이름과 나이가 다른 객체 3개를 만들었다
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

p1 = Person("Emil", 36)
p2 = Person("홍길동", 16)
p3 = Person("이순신", 26)

print(f"{p1.name}의 나이는 {p1.age}살 입니다.")
print(f"{p2.name}의 나이는 {p2.age}살 입니다.")
print(f"{p3.name}의 나이는 {p3.age}살 입니다.")
```

```python
# 매개변수 기본값: age를 넘기지 않으면 18
class Person:
  def __init__(self, name, age=18):
    self.name = name
    self.age = age

p1 = Person("Emil")
p2 = Person("Tobias", 25)

print(p1.name, p1.age)
print(p2.name, p2.age)
```

```python
# 메서드: 클래스 안에 정의한 함수. 첫 매개변수 self로 자기 자신의 속성에 접근한다
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def greet(self):
    print("Hello, my name is " + self.name)

p1 = Person("Emil", 25)
p1.greet()
```

```python
class Person:
  def __init__(self, name):
    self.name = name

  def printname(self):
    print(self.name)

p1 = Person("Tobias")
p2 = Person("Linus")

p1.printname()
p2.printname()
```

```python
# 첫 매개변수 이름은 self가 아니어도 동작한다 (myobject, abc)
# 하지만 관례상 모두 self를 쓰므로 self로 쓰는 게 좋다
class Person:
  def __init__(myobject, name, age):
    myobject.name = name
    myobject.age = age

  def greet(abc):
    print("Hello, my name is " + abc.name)

p1 = Person("Emil", 36)
p1.greet()
```

```python
# 실습: 자동차 정보를 담는 클래스
class Car:
  def __init__(self, brand, model, year):
    self.brand = brand
    self.model = model
    self.year = year

  def display_info(self):
    print(f"{self.year} {self.brand} {self.model}")

car1 = Car("Toyota", "Corolla", 2020)
car1.display_info()
```

```python
# 메서드 안에서 다른 메서드를 self.greet()처럼 불러 쓸 수 있다
class Person:
  def __init__(self, name):
    self.name = name

  def greet(self):
    return "Hello, " + self.name

  def welcome(self):
    message = self.greet()
    print(message + "! Welcome to our website.")

p1 = Person("Tobias")
p1.welcome()
```

</details>
