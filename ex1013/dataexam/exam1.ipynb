### <span style='color:red'>urllib.request.urlopen()</span> 함수로 웹페이지 소스 크롤링하기 

````python
import urllib.request
res = urllib.request.urlopen("http://www.naver.com/")
print(type(res))
print(res.status)
````

````python

print("NAVER 웹페이지의 소스 내용----------------------------------------------------------------")
print(res.read(1000).decode('utf-8'))
````

**📝 정리** — `urlopen()`은 응답 객체(`HTTPResponse`)를 돌려주고 `status`가 200이면 성공이다. `read()`는 **바이트**를 돌려주므로 `decode('utf-8')`로 문자열로 바꿔야 한글이 보인다. `read(1000)`처럼 크기를 주면 앞부분만 읽는다.

````python
import urllib.request
res = urllib.request.urlopen("http://unico2013.dothome.co.kr/crawling/get.php")
print("[ header 정보 ]----------")
res_header = res.getheaders()
for s in res_header :
    print(s)
print("[ body 내용 ]-----------")
print(res.read()) # 바이트 문자열, 바이트열 - b'..............'
````

````python
import urllib.request
res = urllib.request.urlopen("http://unico2013.dothome.co.kr/crawling/get.php")
print("[ header 정보 ]----------")
res_header = res.getheaders()
for s in res_header :
    print(s)
print("[ body 내용 ]-----------")
print(res.read().decode('utf-8'))
````

**📝 정리** — 같은 요청을 `decode()` 없이 출력하면(위) 한글이 `\xec\x9d\xb4...` 같은 바이트로 보이고, `decode('utf-8')`을 하면(아래) 정상적으로 보인다. 응답 헤더의 `Content-Type: text/html; charset=UTF-8`이 어떤 인코딩으로 풀어야 하는지 알려준다.

### urllib.parse.urlencode() 함수로 Query 문자열(요청 파라미터) 인코딩하기

````python
import urllib
print("[ Query문자열 또는 요청 파라미터 인코딩 ]")
params1 = urllib.parse.urlencode({'number': 12524, 'type': 'issue', 'action': 'show'})
print(params1)
````

````python
params2 = urllib.parse.urlencode({'addr': '서울시 강남구 역삼동'})
print(params2)
````

````python
import urllib
print(urllib.parse.quote_plus('가나다 ABC123'))
query_str1 = "q="+urllib.parse.quote_plus('가나 다')
query_str2 = urllib.parse.urlencode({"q" : "가나 다"})
print("-------------------------------------------")
print(query_str1)
print(query_str2)
````

**📝 정리** — URL에는 한글과 공백을 그대로 넣을 수 없어서 인코딩해야 한다. `urlencode()`는 딕셔너리를 `키=값&키=값` 형태로 만들면서 인코딩하고, `quote_plus()`는 문자열 하나를 인코딩한다. 둘 다 공백을 `+`로 바꿔서 결과가 같았다.

### Query 문자열 전달하면서 GET 방식으로 요청하기

````python
import urllib.request
import urllib.parse
params = urllib.parse.urlencode({'name': '유니코', 'age': 10})
print(f"URL 인코딩 규칙이 적용된 문자열 : {params}")
````

````python

url = f"http://unico2013.dothome.co.kr/crawling/get.php?{params}"
print(f"URL 문자열 : {url}")
with urllib.request.urlopen(url) as f:
    print(f.read().decode('utf-8'))
````

### 요청 파라미터 전달하면서 POST 방식으로 요청하기(1) 
**<span style="color:red">urllib.request.urlopen() 사용</span>**

````python
import urllib.request
import urllib.parse
data = urllib.parse.urlencode({'name': '유니코', 'age': 10})
print(f"URL 인코딩 규칙이 적용된 문자열 : {data}")
postdata = data.encode('ascii')
print(f"변환된 바이트 문자열 : {postdata}")
url = "http://unico2013.dothome.co.kr/crawling/post.php"
with urllib.request.urlopen(url, data=postdata) as f:
    print(f.read().decode('utf-8'))
````

### 요청 파라미터 전달하면서 POST 방식으로 요청하기(2)
**<span style="color:red">urllib.request.Request() 와 urllib.request.urlopen() 사용</span>**

````python
import urllib.request
import urllib.parse
data = urllib.parse.urlencode({'name': '유니코', 'age': 10})
postdata = data.encode('ascii')
req = urllib.request.Request(url='http://unico2013.dothome.co.kr/crawling/post.php',
            data=postdata)
print(req)
with urllib.request.urlopen(req) as f:
    print(f.read().decode('utf-8'))
````

**📝 정리** — POST 방식은 파라미터를 URL이 아니라 요청 본문(body)에 담는다. `urlopen()`에 `data=`를 주면 POST로 요청하는데, 이때 데이터는 문자열이 아니라 **바이트**여야 해서 `encode('ascii')`를 했다. `Request` 객체를 먼저 만들어 넘겨도 결과는 같다.

### <span style='color:red'>requests.request()</span> 함수로 웹페이지 소스 크롤링하기 

````python
import requests

r = requests.request('get', 'http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
print('----------------------------------------------------------')
r = requests.request('head', 'http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
print('----------------------------------------------------------')
r = requests.request('post', 'http://unico2013.dothome.co.kr/crawling/post.php', data= {'name':'백도', 'age' : 12})
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
````

````python
import requests

r = requests.get('http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
print('----------------------------------------------------------')
r = requests.head('http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
print('----------------------------------------------------------')
r = requests.post('http://unico2013.dothome.co.kr/crawling/post.php', data= {'name':'백도', 'age' : 12})
print(type(r))
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
````

**📝 정리** — `requests`는 `urllib`보다 간단하다. `requests.request('get', url)`과 `requests.get(url)`은 같은 요청이다. HEAD 요청은 헤더만 받고 본문은 받지 않아서 "응답된 콘텐츠가 없어요"가 출력됐다. POST는 `data=`에 딕셔너리를 그대로 넣으면 된다(인코딩을 알아서 해 줌).

````python
import requests

r = requests.get('http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
print(r.headers)
if r.content :
    print(r.content) 
    print('-'*100)
    print(r.content.decode('utf-8'))
else :
    print('응답된 콘텐츠가 없어요')
````

**📝 정리** — `r.content`는 바이트, `r.text`는 `requests`가 인코딩을 추측해서 바꾼 문자열이다.

````python
import requests

r = requests.get('http://unico2013.dothome.co.kr/crawling/get.php')
print(type(r))
print(r.headers)
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
````

````python
import requests

r = requests.get('http://unico2013.dothome.co.kr/crawling/exam.html')
r.encoding = 'utf-8'
print(type(r))
print(r.headers)
if r.text :
    print(r.text)
else :
    print('응답된 콘텐츠가 없어요')
````

**📝 정리** — 이 페이지는 응답 헤더가 `Content-Type: text/html`로 charset이 없다. 이럴 때 `requests`는 기본값(ISO-8859-1)으로 풀어서 한글이 깨질 수 있어서, `r.encoding = 'utf-8'`로 직접 지정했다.

````python
import requests

dicdata = {'key1': 'value1', 'key2': 'value2'}
urlstr = 'http://unico2013.dothome.co.kr/crawling/get.php'
r = requests.get(urlstr, params=dicdata)
print(r.url)
print('------------------------------------')
dicdata = {'key1': 'value1', 'key2': ['value2', 'value3']}
r = requests.request('GET', urlstr, params=dicdata)
print(r.url)
print('------------------------------------')
tupledata = [('key1', 'value1'), ('key1', 'value2')]
r = requests.get(urlstr, params=tupledata)
print(r.url)
````

**📝 정리** — `params=`에 딕셔너리를 주면 `requests`가 쿼리 문자열을 만들어 준다. 값이 리스트면 같은 키가 여러 번 붙고(`key2=value2&key2=value3`), 튜플 리스트로 주면 같은 키를 여러 번 직접 지정할 수 있다.

````python
import requests

urlstr = 'http://unico2013.dothome.co.kr/crawling/get.php'
r = requests.get(urlstr)
print(r.text)
print('------------------------------------')
print(r.content)
print('------------------------------------')
print(r.content.decode('utf-8'))
````

````python
print(bs.find('p').attrs)
````

**📝 정리** — `.attrs`는 태그의 속성을 딕셔너리로, `.get('align')`은 속성 하나의 값을 돌려준다.

````python
html="""
<!DOCTYPE html>
<html>
  <head>
     <meta charset='utf-8'>
     <title>Test BeautifulSoup</title>
  </head>
  <body>
     <p align="center"> text contents </p>
     <p align="right" class="myp">  text contents 2 </p>
     <p align="left" a="b">   text contents 3 </p>
     <img src="http://unico2013.dothome.co.kr/image/flower.jpg" width="500">
  </body>
</html> """

from bs4 import BeautifulSoup
bs = BeautifulSoup(html, 'html.parser')
print(bs.find('p', align="center"))
print(bs.find('p', class_="myp"))
print(bs.find('p', align="left"))
print("-------------------------------------")
print(bs.find('p', attrs={"align":"center"}))
print(bs.find('p', attrs={"align":"right", "class":"myp"}))
print(bs.find('p', attrs={"align":"left"}))

pdom = bs.find('p', align="left")
print("["+pdom.text+"]")
print("["+pdom.get_text(strip=True)+"]")
print("["+pdom.string.strip()+"]")
````

**📝 정리** — `find()`에 속성 조건을 줄 수 있다. `class`는 파이썬 예약어라서 `class_=`로 쓰거나 `attrs={"class": ...}`로 쓴다.

````python
import requests
from bs4 import BeautifulSoup

req = requests.get('http://unico2013.dothome.co.kr/crawling/exercise_css.html')
html = req.content

bs = BeautifulSoup(html, 'html.parser')
title = bs.select('h1')
title1 = bs.select('#f_subtitle')
title2 = bs.select('.subtitle')
title3 = bs.select('aside > h2')
img = bs.select('[src]')

print("<h1>태그의 갯수 : %d " %len(title))
print("f_subtitle이라는 id 속성을 갖는 태그 갯수 : %d " %len(title1))
print("subtitle이라는 class 속성을 갖는 태그 갯수 : %d " %len(title2))
print("aside 태그의 <h2> 자식 태그 갯수 : %d " %len(title3))
print("src 속성을 갖는 태그 갯수 : %d " %len(img))

for content in title:
    print(content.string)
print("------------------------------")
for content in title1:
    print(content.text)
print("------------------------------")
for content in title2:
    print(content.text)
print("------------------------------")
for content in title3:
    print(content.text)
print("------------------------------")
for content in img:
   print(content["src"])
````

**📝 정리** — `select()`는 CSS 선택자로 찾고 결과는 항상 **리스트**다. `#id`, `.class`, `부모 > 자식`, `[속성]` 선택자를 써 봤다. `subtitle` 클래스는 2개였다.

### YES24 사이트에 올려져 있는 '파이썬' 관련 서적 정보 수집하기

````python
import requests
from bs4 import BeautifulSoup
title = []
link = []
search_book = 'python'
urlstr = f'https://www.yes24.com/Product/Search?domain=BOOK&query={search_book}'
r = requests.get(urlstr)
bs = BeautifulSoup(r.text, 'html.parser')
bookList = bs.select('a.gd_name')

for bookDom in bookList:
    title.append(bookDom.string)
    link.append(bookDom["href"])

print("-- 도서 제목 --")
print(title)
print("-- 도서 링크 URL --")
print(link)
````

**📝 정리** — YES24 검색 결과에서 도서 제목은 `a.gd_name` 태그에 있다. 한 페이지에 24권이 나왔고, 링크는 `/product/goods/번호` 같은 **상대 경로**라서 실제로 쓰려면 `https://www.yes24.com`을 앞에 붙여야 한다.

````python
import requests
from bs4 import BeautifulSoup
title = []
link = []
query_dict = {'domain':'BOOK', 'query':'python'}
urlstr = 'https://www.yes24.com/Product/Search'
r = requests.get(urlstr, params=query_dict)
bs = BeautifulSoup(r.text, 'html.parser')
bookList = bs.select('a.gd_name')

for bookDom in bookList:
    title.append(bookDom.string)
    link.append(bookDom["href"])

print("-- 도서 제목 --")
print(title)
print("-- 도서 링크 URL --")
print(link)

with open('output/booklink_1page.csv', "wt", encoding="utf-8") as f:
    f.write('booktitle,booklink\n')  
    for i in range(len(title)):
        f.write(title[i]+","+link[i]+'\n')

````

**📝 정리** — CSV를 문자열 더하기로 직접 썼다. 1페이지 결과는 제목에 쉼표가 없어서 24행 × 2열로 잘 저장됐지만, 2페이지에 있는 "문제 해결의 언어, 파이선 Python" 같은 제목이 들어가면 열이 하나 더 생겨서 깨진다. 직접 쓸 때는 `csv` 모듈이나 pandas를 쓰는 게 안전하다.

````python
import requests
from bs4 import BeautifulSoup
title = []
link = []
urlstr = 'https://www.yes24.com/Product/Search?domain=BOOK&query=python&page='

for pnum in range(1, 11) :
    urlstr_pagenum = urlstr+str(pnum)
    req = requests.get(urlstr_pagenum)
    print(f'{urlstr_pagenum} 요청.....')
    bs = BeautifulSoup(req.text, 'html.parser')
    book_list = bs.select('a.gd_name')
    
    for dom in book_list:
        print(dom.string)
        title.append(dom.string)
        link.append(dom["href"])

print("-- 도서 제목 --")
print(title)
print("-- 도서 링크 URL --")
print(link)
import pandas as pd
dictdata = {"booktitle":title,"booklink":link}
df = pd.DataFrame(dictdata)
df.to_csv("output/booklink_10page.csv")
````

**📝 정리** — 1~10페이지를 돌며 240권을 모았다. 결과는 pandas `to_csv()`로 저장해서 쉼표가 있는 제목도 따옴표로 감싸져 안전하다. 다만 `to_csv()`는 기본으로 인덱스 열까지 저장해서, 다시 읽으면 `Unnamed: 0` 열이 생긴다. 인덱스가 필요 없으면 `index=False`를 준다. 같은 책("Python으로 시작하는 빅데이터분석 및 인공지능")이 1페이지와 2페이지에 한 번씩 나와 중복이 1건 있었다.

> https://www.yes24.com/Product/Search?domain=BOOK&query=python&page=1 부터
> https://www.yes24.com/Product/Search?domain=BOOK&query=python&page=10 까지

> 제목과 이 제목에 설정된 링크의 URL 문자열을 추출해서 각각
> 리스트에 저장하여 출력하는 파이썬 코드를 작성해줘
> ```

> 🖼️ *(Gemini 답변 화면 캡처 — 저장소에는 넣지 않음)*

````python
import requests
from bs4 import BeautifulSoup
import time

def get_yes24_python_books():
    # 데이터를 저장할 빈 리스트 생성
    all_titles = []
    all_urls = []
    
    # 1페이지부터 10페이지까지 반복
    for page in range(1, 11):
        url = f"https://www.yes24.com/Product/Search?domain=BOOK&query=python&page={page}"
        
        # 웹사이트 차단을 방지하기 위한 헤더 설정
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        headers = {
            "User-Agent": "Mozilla/5.0"
        }
        
        try:
            response = requests.get(url, headers=headers)
            if response.status_code == 200:
                soup = BeautifulSoup(response.text, 'html.parser')
                
                # YES24 검색 결과에서 도서 제목과 링크를 담고 있는 클래스 'gd_name' 선택
                book_elements = soup.select('a.gd_name')
                
                for element in book_elements:
                    title = element.get_text().strip()
                    link = "https://www.yes24.com" + element.get('href')
                    
                    # 각각의 리스트에 추가
                    all_titles.append(title)
                    all_urls.append(link)
                
                print(f"{page}페이지 추출 완료 (현재 누적 도서 수: {len(all_titles)}권)")
            
            # 서버 부하 방지를 위해 0.5초 대기
            time.sleep(0.5)
            
        except Exception as e:
            print(f"{page}페이지 추출 중 오류 발생: {e}")

    return all_titles, all_urls

# 코드 실행
titles, urls = get_yes24_python_books()

# 결과 출력 (데이터가 많으므로 상위 10개만 예시 출력)
print("\n" + "="*50)
print(f"총 추출된 도서 수: {len(titles)}권")
print("="*50)

print("\n[추출된 제목 리스트 (상위 5개)]")
print(titles[:5])

print("\n[추출된 URL 리스트 (상위 5개)]")
print(urls[:5])
````

**📝 정리** — Gemini가 만든 코드. 제목은 `get_text().strip()`으로 꺼내고 링크는 `https://www.yes24.com`을 붙여 **절대 URL**로 만들었다. `User-Agent` 헤더, `try/except`, 페이지마다 `time.sleep(0.5)`로 서버 부담을 줄이는 처리까지 들어 있다. (`headers`를 두 번 정의해서 첫 번째는 바로 덮어써진다.) 결과는 직접 짠 코드와 같은 240권.

### 챗 지피티
> 💬 **ChatGPT에게 보낸 프롬프트** (캡처 이미지 대신 글로 옮김)
> ```
> https://www.yes24.com/Product/Search?domain=BOOK&query=python&page=1 부터
> https://www.yes24.com/Product/Search?domain=BOOK&query=python&page=10 까지
> 각 페이지마다 출력되는 파이썬 관련 서적의
> 제목과 이 제목에 설정된 링크의 URL 문자열을 추출해서 각각
> 리스트에 저장하여 출력하는 파이썬 코드를 작성해줘
> ```

> 🖼️ *(ChatGPT 답변 화면 캡처 — 저장소에는 넣지 않음)*

````python
import requests
from bs4 import BeautifulSoup

# 결과 저장 리스트
titles = []
links = []

# 헤더 (중요: 크롤링 차단 방지)
headers = {
    "User-Agent": "Mozilla/5.0"
}

# 페이지 1 ~ 10
for page in range(1, 11):
    url = f"https://www.yes24.com/Product/Search?domain=BOOK&query=python&page={page}"
    
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.text, "html.parser")
    
    # 책 리스트 영역 찾기 (YES24 구조 기준)
    items = soup.select("div.itemUnit")  # 구조 바뀌면 수정 필요
    
    for item in items:
        title_tag = item.select_one("a.gd_name")
        
        if title_tag:
            title = title_tag.text.strip()
            link = "https://www.yes24.com" + title_tag.get("href")
            
            titles.append(title)
            links.append(link)

# 출력
print("📚 제목 리스트:")
for t in titles:
    print(t)

print("\n🔗 링크 리스트:")
for l in links:
    print(l)
````

**📝 정리** — ChatGPT가 만든 코드. 책 한 권 단위인 `div.itemUnit`을 먼저 찾고 그 안에서 `select_one("a.gd_name")`으로 제목을 찾는 방식이다. 제목 240개, 링크 240개로 결과는 같았다. 같은 프롬프트에도 AI마다 코드 구조가 달랐고, 둘 다 실행해서 결과 개수를 확인해 보는 게 중요했다.
