> **한눈에 보기**
> - `for x in 리스트/문자열:` 원소를 하나씩 꺼내며 반복
> - `break`는 반복을 끝내고, `continue`는 이번 차례만 건너뛴다 (print 앞에 있느냐 뒤에 있느냐로 결과가 달라짐)
> - `range(끝)`, `range(시작, 끝)`, `range(시작, 끝, 간격)` — 끝 숫자는 포함하지 않는다
> - `for … else`: break 없이 끝까지 돌았을 때만 else 실행
> - 중첩 반복, 빈 자리를 채우는 `pass`

### 1. 리스트·문자열 반복

```python
# 리스트의 원소를 하나씩 x에 넣으며 반복
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
# 문자열도 한 글자씩 반복할 수 있다
for x in "banana":
  print(x)
```

### 2. break와 continue

```python
# print 다음에 break -> banana까지 출력하고 멈춤
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
# break가 print보다 먼저 -> banana는 출력되지 않음
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    break
  print(x)
```

```python
# continue: 이번 차례만 건너뛰고 다음 원소로
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

### 3. range()

```python
# range(6): 0부터 5까지 (6은 포함 안 함)
for x in range(6):
  print(x)
```

```python
# range(시작, 끝): 2부터 5까지
for x in range(2, 6):
  print(x)
```

```python
# range(시작, 끝, 간격): 2부터 3씩 늘려 30 전까지
for x in range(2, 30, 3):
  print(x)
```

### 4. for … else

```python
# 반복이 끝까지 돌면 else 실행
for x in range(6):
  print(x)
else:
  print("Finally finished!")
```

```python
# break로 빠져나가면 else는 실행되지 않음
for x in range(6):
  if x == 3: break
  print(x)
else:
  print("Finally finished!")
```

### 5. 중첩 반복과 pass

```python
# 바깥 반복 1번마다 안쪽 반복이 전부 돈다 -> 3 x 3 = 9줄
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

```python
# pass: 아무것도 안 함. 블록이 비어 있으면 문법 오류라서 자리만 채울 때 쓴다
for x in [0, 1, 2]:
  pass
```
