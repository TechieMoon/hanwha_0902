```python
day = 4
match day:
    case 1:
        print("월요일")
    case 2:
        print("화요일")
    case 3:
        print("수요일")
    case 4:
        print("목요일")
    case 5:
        print("금요일")
    case 6:
        print("토요일")
    case 7:
        print("일요일")
```

```python
day = 4
match day:
    case 6:
        print("오늘은 토요일")
    case 7:
        print("오늘은 일요일")
    case _:
        print("주말을 기다리는 중")
```

```python
day = 4
match day:
    case 1 | 2 | 3 | 4 | 5:
        print("오늘은 주중")
    case 6 | 7:
        print("주말이다!")
```
