> **한눈에 보기**
> - `match 값:` 아래 `case`들 중 **같은 값 하나**를 찾아 그 부분만 실행 (break 필요 없음)
> - `case _:`는 어디에도 안 맞을 때(else 역할), `case 1 | 2 | 3:`처럼 여러 값을 묶을 수 있다
> - 파이썬 3.10부터 쓸 수 있다

```python
# day와 같은 case를 위에서부터 찾아 그 부분만 실행
day = 4
match day:
    case 1:
        print("월요일")
    case 2:
        print("화요일")
    case 3:
        print("수요일")
    case 4:
        print("목요일")
    case 5:
        print("금요일")
    case 6:
        print("토요일")
    case 7:
        print("일요일")
```

```python
# case _: 어떤 case에도 맞지 않을 때 (if문의 else 역할)
day = 4
match day:
    case 6:
        print("오늘은 토요일")
    case 7:
        print("오늘은 일요일")
    case _:
        print("주말을 기다리는 중")
```

```python
# | 로 여러 값을 한 case에 묶기
day = 4
match day:
    case 1 | 2 | 3 | 4 | 5:
        print("오늘은 주중")
    case 6 | 7:
        print("주말이다!")
```
