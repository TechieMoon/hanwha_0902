> **한눈에 보기**
> - `while 조건:` 조건이 참인 동안 반복 — **조건을 바꾸는 줄(`i += 1`)을 빼먹으면 무한 반복**
> - `break`로 반복 종료, `continue`로 이번 차례 건너뛰기
> - `while … else`: 조건이 거짓이 되어 끝나면 else 실행 (break로 끝나면 실행 안 됨)

```python
# 조건이 참인 동안 반복. i += 1이 없으면 i가 계속 1이라 무한 반복
i = 1
while i < 6:
    print(i)
    i += 1
```

```python
# i가 3이면 break로 반복 종료
i = 1
while i < 6:
    print(i)
    if i == 3:
        break 
    i += 1
```

```python
# i += 1을 continue보다 먼저 한다 -> 3만 건너뛰고, 무한 반복도 안 생긴다
i = 0 
while i < 6:
    i += 1 
    if i == 3:
        continue 
    print(i)
```

```python
# 조건이 거짓이 되어 반복이 끝나면 else 실행
i = 1
while i < 6:
    print(i) 
    i += 1 
else:
    print("i는 더이상 6보다 작지 않아요")
```
