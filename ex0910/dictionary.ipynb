딕셔너리 생성

```python
myfamily = {
    "child1":{
        "name": "Email",
        "year": 2004
    },
    "child2":{
        "name": "Tobias",
        "year": 2007
    },
    "child3": {
        "name": "Linus",
        "year": 2011
    }
}

print(myfamily)
```

딕셔너리 접근하기

```python
print(myfamily)
print(myfamily["child2"])
print(myfamily["child2"]["name"])
```

딕셔너리 값 수정하기

```python
myfamily["child2"]["name"] = "Alice"
print(myfamily["child2"]["name"])
```

딕셔너리 타입

```python
child1 = "엘리스"
child2 = "조나단" 
child3 = "피터" 
print(child1, child2, child3) 
```

```python
child1 = {"이름": "엘리스", "취미": "배드민턴"}
child2 = {"이름": "조나단", "취미": "피아노"}
child3 = {"이름": "피터", "취미": "영화시청"}
```

```python
yourfamily = {
    "child1": child1,
    "child2": child2,
    "child3": child3
}

print(yourfamily)
```
