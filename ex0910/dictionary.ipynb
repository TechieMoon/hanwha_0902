> **한눈에 보기**
> - 딕셔너리 안에 딕셔너리를 넣는 **중첩 딕셔너리** 만들기
> - 꺼낼 때는 키를 차례로: `myfamily['child2']['name']`, 바꿀 때도 같은 방법으로 `= 새 값`
> - 변수 여러 개 대신 딕셔너리로 묶으면 한 묶음으로 다루기 쉽다

딕셔너리 생성

```python
# 중첩 딕셔너리: 키 -> 또 다른 딕셔너리
myfamily = {
    "child1":{
        "name": "Email",
        "year": 2004
    },
    "child2":{
        "name": "Tobias",
        "year": 2007
    },
    "child3": {
        "name": "Linus",
        "year": 2011
    }
}

print(myfamily)
```

딕셔너리 접근하기

```python
# 바깥 키로 꺼내고, 다시 안쪽 키로 꺼낸다
print(myfamily)
print(myfamily["child2"])
print(myfamily["child2"]["name"])
```

딕셔너리 값 수정하기

```python
# 같은 방법으로 접근해서 = 로 값을 바꾼다
myfamily["child2"]["name"] = "Alice"
print(myfamily["child2"]["name"])
```

딕셔너리 타입

```python
# 비교용: 딕셔너리 없이 변수 3개에 이름만 저장
child1 = "엘리스"
child2 = "조나단" 
child3 = "피터" 
print(child1, child2, child3) 
```

```python
# 아이마다 정보를 딕셔너리로 먼저 만들고
child1 = {"이름": "엘리스", "취미": "배드민턴"}
child2 = {"이름": "조나단", "취미": "피아노"}
child3 = {"이름": "피터", "취미": "영화시청"}
```

```python
# 만들어 둔 딕셔너리 3개를 값으로 넣어 중첩 딕셔너리 완성
yourfamily = {
    "child1": child1,
    "child2": child2,
    "child3": child3
}

print(yourfamily)
```

**📝 정리** — 한 묶음이 되었으므로 `for key, info in yourfamily.items(): print(key, info['이름'])`처럼 반복하기도 쉽다.
