> **한눈에 보기**
> - `if` / `elif` / `else`: 위에서부터 **처음으로 참인 조건 하나만** 실행
> - 실행할 문장이 하나면 한 줄로: `if c < d: print(...)`
> - 조건 표현식(삼항 연산자): `A if 조건 else B`

### 1. if / elif / else

```python
# 조건이 참이면 들여쓴 줄 실행
a = 33 
b = 200 
if b > a: 
    print("b는 a보다 크다")
```

```python
c = 33
d = 200
# d는 c보다 크므로 아래 print문을 실행되지 않는다
if d < c:
    print("c는 d보다 크다")
```

```python
# elif: 앞 조건이 거짓일 때 다음 조건 검사
# 둘 다 거짓이면 아무것도 출력 안 됨 (else가 없으므로)
num1 = 50 
num2 = 100

if num2 > num1:
    print("100은 50보다 큽니다.")
elif num1 == num2:
    print("num1과 num2는 같아요.")
```

```python
# 위에서부터 처음 참인 조건 하나만 실행 -> 75는 'C'
# (score >= 60도 참이지만 이미 C에서 끝남)
score = 75 

if score >= 90:
    print("A") 
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

Short Hand If

```python
# 실행할 문장이 하나면 한 줄로 쓸 수 있다
c = 33
d = 200
if c < d: print("c는 d보다 작다.")
```

```python
# 조건 표현식: (참일 때) if 조건 else (거짓일 때)
a = 100
b = 300
print("a는 b보다 크다") if a > b else print("b는 a보다 크다")
```

## 🔁 복습 질문

<details>
<summary>1. <code>score = 75</code>일 때 90/80/70/60 순서의 elif에서 출력은? 왜 D는 안 나오나?</summary>

`C`. `score >= 60`도 참이지만, if/elif는 처음 참인 조건 하나만 실행하고 끝난다.
</details>

<details>
<summary>2. 조건 순서를 <code>score >= 60</code>부터 쓰면 어떻게 되나?</summary>

90점도 첫 조건에서 걸려 `D`가 된다. 범위 조건은 큰 값(좁은 조건)부터 검사한다.
</details>

<details>
<summary>3. <code>print("A") if a > b else print("B")</code>를 값으로 바꿔 쓰면?</summary>

`print("A" if a > b else "B")`. 조건 표현식은 값을 고를 때 쓰는 것이 일반적이다.
</details>
