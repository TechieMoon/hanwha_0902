```python
a = 33 
b = 200 
if b > a: 
    print("b는 a보다 크다")
```

```python
c = 33
d = 200
# d는 c보다 크므로 아래 print문을 실행되지 않는다
if d < c:
    print("c는 d보다 크다")
```

```python
num1 = 50 
num2 = 100

if num2 > num1:
    print("100은 50보다 큽니다.")
elif num1 == num2:
    print("num1과 num2는 같아요.")
```

```python
score = 75 

if score >= 90:
    print("A") 
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

Short Hand If

```python
c = 33
d = 200
if c < d: print("c는 d보다 작다.")
```

```python
a = 100
b = 300
print("a는 b보다 크다") if a > b else print("b는 a보다 크다")
```
