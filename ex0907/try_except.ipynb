> **한눈에 보기**
> - `try`에서 오류가 나면 프로그램이 멈추지 않고 `except`로 넘어간다
> - `except NameError:`처럼 **오류 종류별로** 나눠 잡을 수 있다 (위에서부터 맞는 것 하나만 실행)
> - `else`: 오류가 없을 때만 실행 / `finally`: 오류가 있든 없든 마지막에 항상 실행

### 1. try / except 기본

```python
# x를 만든 적이 없으므로 print(x)에서 NameError -> except로 이동
try:
    print(x)
except:
    print("An exception occurred")
```

x가 정의되지 않았으므로
try에서 에러가 발생 -> 프로그램이 종료되지 않고 except문을 실행
따라서 "An exception occurred"가 출력됨

```python
# except 오류이름: 그 종류의 오류만 잡는다. 위에서부터 차례로 맞는 것을 찾는다
try:
  print(x)
except NameError:
  print("Variable x is not defined")
except:
  print("Something else went wrong")
```

Error도 종류가 있음
NameError가 발생했으므로 NameError가 있는 except가 실행됨

### 2. else와 finally

```python
# else: try에서 오류가 없을 때만 실행
try:
  print("Hello")
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
```

try가 에러가 발생하지 않으면 else 실행

```python
# x가 없어 오류 -> except 실행, else는 건너뜀
try:
  print(x)
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
```

try가 에러가 발생하면 except 실행하고 else는 실행 안함

```python
# 바깥 try: 파일 열기 실패를 잡는다 (demofile.txt가 없어서 여기서 실패)
# 안쪽 try: 쓰기 실패를 잡고, finally로 무슨 일이 있어도 파일을 닫는다
# 참고: open()은 기본이 읽기 모드('r')라서, 파일이 있었다면 write에서 실패했을 것이다 (쓰려면 open(..., 'w'))
try:
  f = open("demofile.txt")
  try:
    f.write("Lorum Ipsum")
  except:
    print("Something went wrong when writing to the file")
  finally:
    f.close()
except:
  print("Something went wrong when opening the file")
```

finally 구문은 try가 잘 실행되든 error가 발생하든 마지막에 실행
파일은 사용하고 꼭 닫아야한다.

**📝 정리** — `except:`만 쓰면 모든 오류를 잡아서 진짜 버그까지 숨길 수 있다. 가능하면 `except NameError:`처럼 종류를 적는다. 파일은 요즘 `with open(...) as f:`로 열면 블록이 끝날 때 자동으로 닫힌다.

<summary>1. <code>try</code> 안에서 오류가 나면 그 아래 줄은 실행되나?</summary>

아니다. 오류가 난 줄에서 바로 `except`로 넘어간다.
</details>

<details>
<summary>2. <code>except NameError:</code>와 <code>except:</code>를 같이 쓰면 NameError일 때 어느 것이 실행되나?</summary>

위에서부터 맞는 것 하나만 실행되므로 `except NameError:`만 실행된다.
</details>

<details>
<summary>3. <code>else</code>와 <code>finally</code>는 각각 언제 실행되나?</summary>

`else`는 `try`에서 오류가 없을 때만, `finally`는 오류 여부와 상관없이 항상 마지막에 실행된다.
</details>

<details>
<summary>4. 파일 작업에 <code>finally</code>를 쓰는 이유는?</summary>

쓰기 중에 오류가 나도 `f.close()`로 파일을 꼭 닫기 위해서다. `with open(...) as f:`를 쓰면 자동으로 닫힌다.
</details>
