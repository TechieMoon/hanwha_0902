```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model="gpt-4o-mini"
)
```

## logprob 활성화

- 모델의 답변 확신도가 0에 가까울수록 정다벵 가까운 확신을 가진다.

```python
# Obj 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini"
).bind(logprobs=True)
```

```python
question2 = "인도의 수도는 어디인가요?"

response = llm_with_logprob.invoke(question2)
```

```python
response
```

```python
response.response_metadata
```

```python
response.response_metadata["logprobs"]
```

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")

for token in answer:
    print(token.content, end="", flush=True)
```
