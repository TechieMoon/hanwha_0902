```python
# dotenv 설치 (정식 패키지 이름은 python-dotenv, 처음 한 번만)
!pip install dotenv
```

```python
# .env 파일을 읽어 환경변수로 등록 (성공하면 True)
from dotenv import load_dotenv

load_dotenv()
```

```python
# 키가 들어왔는지 앞 8글자만 확인 - 키 전체는 절대 출력하지 않는다 (출력도 깃허브에 올라감)
import os 

print(f"키 값은: {os.environ['OPENAI_API_KEY'][:8]}")
```

## 🔁 복습 질문

<details>
<summary>1. <code>load_dotenv()</code>가 <code>True</code>를 돌려주면 무슨 뜻인가?</summary>

`.env` 파일을 찾아서 환경변수로 등록했다는 뜻이다. 그 뒤 `os.environ["OPENAI_API_KEY"]`로 값을 읽을 수 있다.
</details>

<summary>2. <code>!pip --version</code> 결과에서 무엇을 확인하나?</summary>

</details>
