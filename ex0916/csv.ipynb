## 2. 딕셔너리로 DataFrame 만들기

`DataFrame`은 엑셀 시트처럼 행과 열이 있는 2차원 표 자료구조다. 딕셔너리를 넘기면 **key가 컬럼명, value 리스트가 각 컬럼의 값들**이 된다.

```python
# 딕셔너리의 key -> 컬럼명, value(리스트) -> 그 컬럼의 값들이 된다.
mydataset = {
  'cars': ["BMW", "Volvo", "Ford"],
  'passings': [3, 7, 2]
}

myvar = pd.DataFrame(mydataset)

print(myvar)
```

## 3. 리스트로 Series 만들기 (기본 정수 인덱스)

`Series`는 컬럼 하나에 해당하는 1차원 자료구조다. 리스트만 넘기면 인덱스가 자동으로 0부터 매겨진다.

```python
a = [1, 7, 2]
# 인덱스를 따로 지정하지 않으면 0, 1, 2, ... 순서로 자동 부여된다.
myvar1 = pd.Series(a)
print(myvar1)
```

정수 인덱스로 특정 위치의 값을 꺼낼 수 있다 (리스트처럼 0부터 시작).

```python
print(myvar1[1])  # 인덱스 1번(두 번째) 값을 출력
```

## 4. 인덱스를 직접 지정한 Series

`index` 파라미터로 각 값에 이름표를 붙일 수 있다. 이후에는 숫자 대신 이 이름으로 값을 조회할 수 있다.

```python
# index를 직접 지정하면 각 값에 원하는 이름표를 붙일 수 있다.
myvar2 = pd.Series(a, index = ["x", "y", "z"])
print(myvar2)
```

지정한 라벨(`"y"`)로 값을 조회한다.

```python
print(myvar2["y"])  # 라벨 "y"에 해당하는 값 조회
```

## 5. 딕셔너리로 Series 만들기

딕셔너리를 `Series`에 넘기면 **key가 그대로 인덱스(라벨)** 가 된다.

```python
# 딕셔너리를 Series로 만들면 key가 인덱스(라벨)로 그대로 쓰인다.
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar3 = pd.Series(calories)

print(myvar3)
```

## 6. 여러 컬럼을 가진 DataFrame

컬럼이 2개 이상인 딕셔너리를 넘기면 그대로 여러 컬럼을 가진 표가 만들어진다.

```python
# 컬럼이 여러 개인 딕셔너리 -> 여러 컬럼을 가진 DataFrame이 된다.
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

myvar4 = pd.DataFrame(data)

print(myvar4)
```

## 7. 사람 정보로 DataFrame 만들기

이름, 나이, 성별 3개 컬럼을 가진 작은 DataFrame을 만든다. 이후 Titanic 데이터를 다루기 전, DataFrame의 기본 동작(컬럼 선택, 통계 함수 등)을 먼저 연습해본다.

```python
df = pd.DataFrame(
    {
        "Name": [
            "Braund, Mr. Owen Harris",
            "Allen, Mr. William Henry",
            "Bonnell, Miss Elizabeth",
        ],
        "Age": [22, 35, 58],
        "Sex": ["male", "male", "female"],
    }
)
```

만든 DataFrame 전체를 출력해서 확인한다.

```python
print(df)
```

`df["컬럼명"]`으로 특정 컬럼 하나만 선택할 수 있다. 결과는 `Series`다.

```python
print(df["Age"])  # "Age" 컬럼만 선택 -> Series로 반환됨
```

`name` 파라미터로 Series에 이름을 붙일 수도 있다 (DataFrame의 컬럼처럼 취급하기 좋음).

```python
ages = pd.Series([22,35,58], name="Age")
print(ages)
```

Series/컬럼에는 `.max()`, `.min()`, `.mean()` 같은 통계 메서드를 바로 쓸 수 있다.

```python
df["Age"].max()  # Age 컬럼의 최댓값
```

`.describe()`는 숫자형 컬럼들에 대한 요약 통계(개수, 평균, 표준편차, 최소/최대, 사분위수 등)를 한 번에 보여준다.

```python
df.describe()  # 숫자형 컬럼들에 대한 요약 통계량 한 번에 확인
```

## 12. 조건으로 행 필터링하기

`titanic["Age"] > 35`처럼 조건식을 대괄호 안에 넣으면, 그 조건을 만족하는 행만 걸러낼 수 있다. (SQL의 `WHERE`와 비슷한 역할)

```python
# 조건식을 대괄호에 넣으면 조건을 만족하는 행만 걸러낸 DataFrame을 반환한다.
above_35 = titanic[titanic["Age"] > 35]
above_35.head()
```

조건식 `titanic["Age"] > 35` 자체는 각 행이 조건을 만족하는지(True/False)를 담은 **불리언 Series**다. 이 불리언 Series를 다시 대괄호에 넣는 것이 필터링의 원리다.

```python
titanic["Age"] > 35  # 각 행마다 조건 만족 여부(True/False)를 담은 불리언 Series
```

필터링된 결과(`above_35`)의 크기를 확인해서 몇 명이 조건을 만족했는지 본다.

```python
above_35.shape  # 35세 초과 승객의 (행 개수, 열 개수)
```

`|`(OR)로 여러 조건을 결합할 수도 있다. 여기서는 `Pclass`(선실 등급)가 2등석 **또는** 3등석인 승객만 필터링한다. 각 조건은 반드시 괄호로 감싸야 한다 (파이썬 연산자 우선순위 때문).

```python
# | (OR)로 여러 조건 결합. 각 조건은 괄호로 감싸야 한다.
class_23 = titanic[(titanic["Pclass"] == 2) | (titanic["Pclass"] == 3)]
class_23.head()
```

## 13. loc / iloc로 행과 열을 동시에 선택하기

`.loc[행조건, 열이름]`은 **라벨(이름) 기준**으로 행과 열을 동시에 선택한다. 여기서는 "나이가 35세 초과인 행"의 "Name" 컬럼만 뽑아낸다.

```python
# .loc[행조건, 열이름] : 라벨 기준으로 행/열을 동시에 선택.
# "Age > 35"인 행들 중에서 "Name" 컬럼만 뽑는다.
adult_names = titanic.loc[titanic["Age"] > 35, "Name"]
adult_names.head()
```

`.iloc[행범위, 열범위]`는 **정수 위치(순서) 기준**으로 선택한다. 여기서는 10번째~24번째 행(`9:25`)의, 3번째~5번째 열(`2:5`)을 잘라낸다. 슬라이싱이라 끝 인덱스는 포함되지 않는다.

```python
# .iloc[행범위, 열범위] : 정수 위치(순서) 기준으로 선택.
# 인덱스 9~24번 행(10~25번째), 2~4번 열(3~5번째)을 잘라낸다.
titanic.iloc[9:25, 2:5]
```
