> **한눈에 보기**
> - OpenAI 파이썬 SDK로 첫 API 호출 (교재 90쪽)
> - `OpenAI()`는 환경변수 `OPENAI_API_KEY`를 읽는다 — 코드에 키를 쓰지 않는다
> - `client.responses.create(model=..., input=...)` → `response.output_text`로 답만 꺼낸다
> - 정리 문서: [OpenAI가이드.md](OpenAI가이드.md)

```python
# OpenAI 공식 파이썬 SDK
from openai import OpenAI
```

```python
# OPENAI_API_KEY 환경변수를 자동으로 읽는다
# Responses API: model=사용할 모델, input=질문
client = OpenAI()

response = client.responses.create(
    model="gpt-5.6-luna",
    input="안녕하세요라고 한마디만 해줘."
)
```

```python
# 답 글자만 꺼내기
print(response.output_text)
```

```python
# 응답 객체 전체: id, model, output(답 메시지), usage(토큰 사용량) 등이 들어 있다
print(response)
```

**📝 정리** — 출력 끝쪽 `usage`를 보면 입력 16토큰 + 출력 7토큰 = 23토큰을 썼다. API 요금은 이 토큰 수로 계산된다.
