### 2. 변수와 자료형

```python
# 형 변환(casting): str(), int(), float()로 원하는 자료형의 값을 만든다
# float(3)은 3.0으로 출력된다
x = str(3)
y = int(3)
z = float(3)

print(x)
print(y)
print(z)
```

```python
# type(): 변수의 자료형 확인 (int: 정수, str: 문자열)
x = 5
y = "John"
print(type(x))
print(type(y))
```

### 3. 여러 값 출력하고 연결하기

```python
# print에 쉼표로 여러 값을 넘기면 사이에 공백이 하나씩 들어간다
x = "Python"
y = "is"
z = "awesome"
print(x,y,z)
```

```python
# 숫자끼리 + 는 덧셈
x = 10
y = 20
print(x + y)
```

```python
# 변수에 값만 넣는 셀이라 출력이 없다
x = "Python"
y = "is"
z = "awesome"
```

```python
# 숫자와 문자열은 + 로 바로 붙일 수 없다 (str + int -> TypeError)
# str(x)로 숫자를 문자열로 바꾼 뒤 연결한다
x = 5
y = "John"
print(str(x) + y)
```

### 4. 지역 변수와 전역 변수

```python
# 함수 안의 x는 지역 변수라서 함수 밖의 x(전역 변수)와 다른 변수다
# 그래서 함수가 끝난 뒤 밖의 x는 여전히 'awesome'
x = "awesome"

def myfunc():
  x = "fantastic"
  print("Python is " + x)

myfunc()

print("Python is " + x)
```

```python
# global x: 함수 안에서 전역 변수 x를 직접 바꾼다고 선언
def myfunc():
  global x
  x = "fantastic"

myfunc()

print("Python is " + x)
```

### 5. 여러 줄 문자열

```python
# 따옴표 세 개(""")로 줄바꿈이 들어간 문자열을 만들고, +로 다른 문자열과 이어 붙인다
x = """abcde
fgh
d""" + 'fff'
print(x)
```

<details>
<summary>3. 함수 안에서 <code>x = "fantastic"</code>로 바꿨는데 함수 밖의 <code>x</code>가 그대로인 이유는?</summary>

함수 안에서 만든 <code>x</code>는 지역 변수라 밖의 전역 변수 <code>x</code>와 다른 변수이기 때문이다. 밖의 변수를 바꾸려면 함수 안에서 <code>global x</code>를 선언한다.
</details>
